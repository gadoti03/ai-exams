```python
from datetime import date
import subprocess

import ipywidgets as widgets
from IPython.display import display, clear_output

# Text input to enter exam name
exam_name_text = widgets.Text(
    value='Artificial Intelligence Exam',
    description='Exam Name:',
    style={'description_width': '140px'},
    layout=widgets.Layout(width='350px')
)

# Date picker to select exam date
exam_date_picker = widgets.DatePicker(
    value=date.today(),
    description='Exam Date:',
    style={'description_width': '140px'},
    layout=widgets.Layout(width='300px')
)

# Slider to choose how many exams to generate
n_exams_slider = widgets.IntSlider(
    value=5,
    min=1,
    max=20,
    step=1,
    description='Number of exams:',
    style={'description_width': '140px'},
    continuous_update=False,
    layout=widgets.Layout(width='350px')
)

# Dropdown to choose difficulty level
difficulty_dropdown = widgets.Dropdown(
    options=[
        ('Random', 'random'),
        ('Easy (3-4 expansions)', 'easy'),
        ('Medium (5-7 expansions)', 'medium'),
        ('High (8-11 expansions)', 'high')
    ],
    value='random',
    description='Difficulty:',
    style={'description_width': '140px'},
    layout=widgets.Layout(width='350px')
)

# Dropdown to choose consistency
consistency_dropdown = widgets.Dropdown(
    options=[
        ('Yes', 'y'),
        ('No', 'n'),
        ('Mixed', 'm')
    ],
    value='m',
    description='Consistency:',
    style={'description_width': '140px'},
    layout=widgets.Layout(width='350px')
)

# Button to trigger the generation
generate_button = widgets.Button(
    description='Generate Exams',
    button_style='success',
    layout=widgets.Layout(width='120px')
)

# Output area to show messages
output_area = widgets.Output()

# Function to execute when button is clicked
def on_generate_clicked(b):
    clear_output(wait=True)  # Pulisce solo la cella (i log veri), NON l'output grafico

    # Recupera i dati del form
    exam_name = exam_name_text.value
    exam_date = exam_date_picker.value.strftime('%d-%m-%Y') if exam_date_picker.value else date.today().strftime('%d-%m-%Y')
    n_exams = n_exams_slider.value
    difficulty = difficulty_dropdown.value
    consistency = consistency_dropdown.value

    # Stampa SOLO nei log veri (NON nell'output grafico)
    print(f"Exam name: {exam_name}")
    print(f"Exam date: {exam_date}")
    print(f"Generating {n_exams} exams...")
    print(f"Selected difficulty: {difficulty}")
    print(f"Selected consistency: {consistency}")

    # Costruisci il comando
    command = [
        "python3", "generate_exams.py",
        "--n_exams", str(n_exams),
        "--difficulty", difficulty,
        "--exam", exam_name,
        "--date", exam_date,
        "--consistency", consistency
    ]

    print(f"\nRunning command: {' '.join(command)}\n")

    try:
        # Lancia il file esterno
        result = subprocess.run(command, capture_output=True, text=True, cwd="../scripts")
        print(result.stdout)
        if result.stderr:
            print("Errors:", result.stderr)
    except Exception as e:
        print("Failed to run the script:", e)

generate_button.on_click(on_generate_clicked)

# Layout organization
form = widgets.VBox([
    widgets.HTML("<h2>Exam Generator</h2>"),
    widgets.HBox([exam_name_text, exam_date_picker]),
    n_exams_slider,
    difficulty_dropdown,
    consistency_dropdown,
    generate_button,
    output_area
])

display(form)
```
